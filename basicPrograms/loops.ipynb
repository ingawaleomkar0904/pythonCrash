For Loop

```python
range(10)
```

```python
for i in range(10):
    print(i)
```

```python
for i in range(1,6):
    print(i)
```

```python
## Using Step parameter
for i in range(1,10,2):   ## range has 3 parameters (start, stop,step)
    print(i)
```

```python
## Reverse Order
for i in range(10,1,-1):
    print(i)
```

```python
## With Strings

name="Anay"
for i in name:
    print(i)
```

While Loop



```python
## The while loop continues to execute until condition is getting true
count=0
while count<10:
    print(count)
    count+=1
```

Loop Control Statements

```python
## 'break' Statement : The 'break statement' exits the loop prematurely.
for i in range(10):
    if i==5:
        break
    print(i)
```

```python
## continue
## ''continue'' statement skips the current iteration and continues with the next.
for i in range(10):
    if i%2==0:
        continue
    print(i)
```

```python
## 'pass' Statement
## pass statement is a null operation and it does nothing.
for i in range(10):
    if i==4:
        print("We are passing this step",i)
        pass
    print(i)
```

Nested Loops


```python
for i in range(5):
    for j in range(3):
        print(f"i--> {i} and j--> {j}")
```

```python
## Calculate the sum of first N natural numbers using while loop
n=10
addition=0
count=0

while count<=n:
    addition+=count
    count+=1

print("Sum of  first 10 Natural numbers is=",addition)
```

```python
## Calculate the sum of first N natural numbers using for loop

addition=0
for i in range(0,11):
    addition+=i

print("Sum of  first 10 Natural numbers is=",addition)
```
