```python
## Arithmatic Operations

x=float(input("Enter the first number: "))
y=float(input("Enter the second number: "))

add=x+y
sub=x-y
mul=x*y
div=x/y
floordiv=x//y
exp=x**y
mod=x%y
divmod=x//y, x%y

print("Addition--> ",add)
print("Subtraction--> ",sub)
print("Multiplication--> ",mul)
print("Division--> ",div)
print("Floor Division--> ",floordiv)
print("Exponent--> ",exp)
print("Modulus--> ",mod)
print("Div Mod--> ",divmod)
```
