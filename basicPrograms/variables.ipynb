```python
## Simple Basic Arithmatic operations
num1=float(input("Enter the first number:"))
num2=float(input("Enter the second number:"))

add=num1+num2
print("Addition-->",add ,type(add))
subtract=num1-num2
print("Subtraction-->",subtract,type(subtract))
multiply=num1*num2
print("Multiplication-->",multiply,type(multiply))
divide=num1/num2
print("Division-->",divide,type(divide))
```
